```python
import numpy as np
import matplotlib.pyplot as plt

from scipy import signal
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

import tensorflow as tf

print("Bibliotecas cargadas correctamente")
print("TensorFlow:", tf.__version__)
```

```python
frecuencia_muestreo = 8000
frecuencia_portadora = 1000
tasa_simbolos = 1000

muestras_simbolo = int(
    frecuencia_muestreo / tasa_simbolos
)

numero_simbolos = 16

total_muestras = (
    muestras_simbolo * numero_simbolos
)

print("Frecuencia de muestreo:", frecuencia_muestreo)
print("Frecuencia portadora:", frecuencia_portadora)
print("Tasa de símbolos:", tasa_simbolos)
print("Muestras por señal:", total_muestras)
```

```python
def generar_ASK():
    bits = np.random.randint(0, 2, numero_simbolos)

    amplitudes = np.where(bits == 0, 0.3, 1.0)

    amplitudes = np.repeat(amplitudes, muestras_simbolo)

    t = np.arange(total_muestras) / frecuencia_muestreo

    senal = amplitudes * np.cos(
        2 * np.pi * frecuencia_portadora * t
    )

    return senal.astype(np.float32)
```

```python
def generar_FSK():
    bits = np.random.randint(0, 2, numero_simbolos)

    frecuencias = np.where(bits == 0, 800, 1200)

    frecuencias = np.repeat(frecuencias, muestras_simbolo)

    t = np.arange(total_muestras) / frecuencia_muestreo

    fase = np.zeros(total_muestras)

    fase[0] = 0

    for i in range(1, total_muestras):
        fase[i] = fase[i-1] + (
            2 * np.pi * frecuencias[i] / frecuencia_muestreo
        )

    senal = np.cos(fase)

    return senal.astype(np.float32)
```

```python
def generar_PSK():
    bits = np.random.randint(0, 2, numero_simbolos)

    fases = np.where(bits == 0, 0, np.pi)

    fases = np.repeat(fases, muestras_simbolo)

    t = np.arange(total_muestras) / frecuencia_muestreo

    senal = np.cos(
        2 * np.pi * frecuencia_portadora * t + fases
    )

    return senal.astype(np.float32)
```

```python
def generar_QAM():
    simbolos = np.random.randint(0, 16, numero_simbolos)

    niveles = np.array([-3, -1, 1, 3])

    I = niveles[simbolos // 4]
    Q = niveles[simbolos % 4]

    I = np.repeat(I, muestras_simbolo) / np.sqrt(10)
    Q = np.repeat(Q, muestras_simbolo) / np.sqrt(10)

    t = np.arange(total_muestras) / frecuencia_muestreo

    portadora_I = np.cos(
        2 * np.pi * frecuencia_portadora * t
    )

    portadora_Q = np.sin(
        2 * np.pi * frecuencia_portadora * t
    )

    senal = I * portadora_I - Q * portadora_Q

    return senal.astype(np.float32)
```

```python
senal_ask = generar_ASK()
senal_fsk = generar_FSK()
senal_psk = generar_PSK()
senal_qam = generar_QAM()

print("ASK:", senal_ask.shape)
print("FSK:", senal_fsk.shape)
print("PSK:", senal_psk.shape)
print("QAM:", senal_qam.shape)

print("\n¡Los cuatro generadores funcionan correctamente!")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

senales = [
    (senal_ask, "Modulación ASK"),
    (senal_fsk, "Modulación FSK"),
    (senal_psk, "Modulación PSK"),
    (senal_qam, "Modulación 16-QAM")
]

for i, (senal, titulo) in enumerate(senales):

    plt.subplot(4, 1, i + 1)

    plt.plot(senal)

    plt.title(titulo)
    plt.xlabel("Muestras")
    plt.ylabel("Amplitud")

    plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
def agregar_ruido_AWGN(senal, snr_db):

    potencia_senal = np.mean(senal ** 2)

    snr_lineal = 10 ** (snr_db / 10)

    potencia_ruido = potencia_senal / snr_lineal

    ruido = np.random.normal(
        0,
        np.sqrt(potencia_ruido),
        len(senal)
    )

    senal_ruidosa = senal + ruido

    return senal_ruidosa.astype(np.float32)
```

```python
# Número de señales por modulación
muestras_por_modulacion = 1000

# Niveles de ruido que utilizaremos
niveles_snr = [-5, 0, 5, 10, 15, 20]

# Almacenar señales y etiquetas
datos = []
etiquetas = []
niveles_utilizados = []

# Generadores de señales
generadores = {
    "ASK": generar_ASK,
    "FSK": generar_FSK,
    "PSK": generar_PSK,
    "QAM": generar_QAM
}

# Generar las señales
for nombre, generador in generadores.items():

    for i in range(muestras_por_modulacion):

        # Generar señal original
        senal = generador()

        # Seleccionar un nivel de ruido aleatorio
        snr = np.random.choice(niveles_snr)

        # Agregar ruido
        senal_ruidosa = agregar_ruido_AWGN(
            senal, snr
        )

        # Guardar señal y etiqueta
        datos.append(senal_ruidosa)
        etiquetas.append(nombre)
        niveles_utilizados.append(snr)

# Convertir listas en arreglos NumPy
X = np.array(datos, dtype=np.float32)
y = np.array(etiquetas)
snr_datos = np.array(niveles_utilizados)

print("¡Conjunto de datos generado correctamente!")
print("Total de señales:", len(X))
print("Forma de los datos:", X.shape)
print("Total de etiquetas:", len(y))
```

```python
from collections import Counter

conteo = Counter(y)

print("Cantidad de señales por modulación:\n")

for nombre, cantidad in conteo.items():
    print(nombre, ":", cantidad)
```

```python
# Separar datos para entrenamiento y prueba

X_entrenamiento, X_prueba, y_entrenamiento, y_prueba, snr_entrenamiento, snr_prueba = train_test_split(
    X,
    y,
    snr_datos,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Separación de datos completada\n")

print("Datos de entrenamiento:", len(X_entrenamiento))
print("Etiquetas de entrenamiento:", len(y_entrenamiento))

print("Datos de prueba:", len(X_prueba))
print("Etiquetas de prueba:", len(y_prueba))
```

2.1. Investigación de arquitecturas CNN
Las Redes Neuronales Convolucionales (CNN) son modelos de aprendizaje profundo que permiten identificar patrones y características importantes dentro de los datos. Aunque se utilizan principalmente en el procesamiento de imágenes, también pueden aplicarse al análisis de señales y series de tiempo.

En el análisis de señales, las CNN son útiles porque pueden detectar características como cambios de amplitud, frecuencia, variaciones y patrones que permiten diferenciar una señal de otra. Su principal ventaja es que pueden aprender automáticamente estas características durante el entrenamiento, sin necesidad de definir manualmente cada una.

Existen dos formas principales de utilizar CNN para analizar señales:

CNN 1D (Conv1D): Se utiliza directamente sobre señales temporales. Las convoluciones recorren la señal y detectan patrones a lo largo del tiempo. Es una buena alternativa cuando se trabaja con señales representadas mediante valores numéricos, como las señales de comunicación.

CNN 2D (Conv2D): Se utiliza cuando las señales se transforman en imágenes, por ejemplo, mediante espectrogramas. Un espectrograma representa visualmente cómo cambia la frecuencia de una señal a lo largo del tiempo. Esto permite que la red neuronal identifique características visuales que ayudan a distinguir diferentes modulaciones.

Para este proyecto se propone utilizar una arquitectura CNN 1D con capas Conv1D, debido a que permite trabajar directamente con las muestras de las señales y reconocer sus características temporales.
2.2. Diseño del modelo
Para construir el modelo utilizaremos Keras, una biblioteca de aprendizaje profundo que permite crear redes neuronales mediante diferentes capas.

La arquitectura estará formada por:
- Capa de entrada: Recibe los datos de la señal.
- Primera capa Conv1D: Detecta patrones básicos dentro de la señal.
- Segunda capa Conv1D: Identifica características más complejas.
- Activación ReLU: Permite introducir relaciones no lineales en la red.
- Capa Flatten: Convierte los resultados en un vector para conectarlos con las capas finales.
- Capa Dense: Procesa las características aprendidas.
- Capa de salida Softmax: Contiene cuatro neuronas, una para cada categoría de modulación, y calcula la probabilidad de pertenencia a cada una.

```python
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical

# Preparar las señales
X_entrenamiento_CNN = X_entrenamiento.reshape(-1, 128, 1)
X_prueba_CNN = X_prueba.reshape(-1, 128, 1)

# Convertir las etiquetas de texto a números
codificador = LabelEncoder()
y_entrenamiento_num = codificador.fit_transform(y_entrenamiento)
y_prueba_num = codificador.transform(y_prueba)

# Convertir las etiquetas a formato categórico
y_entrenamiento_CNN = to_categorical(y_entrenamiento_num, num_classes=4)
y_prueba_CNN = to_categorical(y_prueba_num, num_classes=4)

print("Datos preparados correctamente")
print("Forma de entrenamiento:", X_entrenamiento_CNN.shape)
print("Forma de prueba:", X_prueba_CNN.shape)
print("Clases:", codificador.classes_)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv1D, Flatten, Dense

print("Capas de la red neuronal importadas correctamente")

modelo = Sequential([
    # Capa de entrada
    Input(shape=(128, 1)),

    # Primera capa convolucional
    Conv1D(filters=32, kernel_size=3, activation='relu'),

    # Segunda capa convolucional
    Conv1D(filters=64, kernel_size=3, activation='relu'),

    # Convertir características en un vector
    Flatten(),

    # Capa de salida: 4 modulaciones
    Dense(4, activation='softmax')
])

print("Modelo CNN creado correctamente")
modelo.summary()
```

```python
modelo.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Modelo compilado correctamente")
```

```python
historial = modelo.fit(
    X_entrenamiento_CNN,
    y_entrenamiento_CNN,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

# Gráfica de precisión
plt.subplot(1, 2, 1)
plt.plot(historial.history['accuracy'], label='Entrenamiento')
plt.plot(historial.history['val_accuracy'], label='Validación')
plt.title('Precisión del modelo')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# Gráfica de pérdida
plt.subplot(1, 2, 2)
plt.plot(historial.history['loss'], label='Entrenamiento')
plt.plot(historial.history['val_loss'], label='Validación')
plt.title('Pérdida del modelo')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

3.1. ¿Confunde la IA la modulación ASK con PSK? ¿Por qué crees que sucede?
Sí, en escenarios con niveles de ruido elevados (como SNR de 0 dB o -5 dB) el modelo puede llegar a confundir ASK con PSK, aunque con SNR alto la distinción es mucho más clara. En mi código, ASK varía la amplitud entre 0.3 y 1.0 manteniendo la fase constante, mientras que PSK mantiene una amplitud constante de 1.0 y cambia la fase entre 0 y π radianes. Cuando el ruido AWGN se suma a la señal, altera aleatoriamente la envolvente y provoca cambios bruscos o cruces irregulares en la onda; de este modo, un salto de fase en PSK puede verse afectado de forma que la red convolucional lo confunda con un cambio de amplitud propio de ASK, y viceversa.
3.3. ¿En qué punto el modelo se vuelve "ciego" al ruido?
El modelo comienza a perder precisión de forma notable en los niveles más bajos de SNR (alrededor de 0 dB y especialmente en -5 dB), acercándose progresivamente al rendimiento por azar (25 % en una clasificación de 4 clases). En -5 dB, la potencia del ruido AWGN supera a la potencia de la señal, lo que enmascara los patrones temporales de amplitud, frecuencia y fase en las 128 muestras de entrada y hace que la red Conv1D ya no pueda distinguir con confiabilidad entre las modulaciones.
3.4. Análisis de errores e hipótesis técnica
Modulación más difícil de clasificar: 16-QAM (y en niveles de ruido altos también se ven afectadas las modulaciones basadas en amplitud como ASK).
Hipótesis técnica: La modulación 16-QAM transmite 4 bits por símbolo combinando 4 niveles en la componente en fase ($I$) y 4 niveles en cuadratura ($Q$), lo que genera una constelación con 16 estados posibles. Al haber más puntos dentro de un mismo rango de potencia, la separación entre símbolos vecinos es menor que en esquemas más simples; por ello, cuando se añade ruido AWGN fuerte (SNR ≤ 0 dB), las variaciones aleatorias desplazan fácilmente la señal hacia regiones cercanas a otros símbolos o alteran su envolvente temporal de manera similar a una señal ASK. Además, la red Conv1D recibe una señal temporal de un solo canal de 128 muestras en lugar de separar explícitamente las componentes $I$ y $Q$, lo que dificulta aún más la separación de los estados en condiciones ruidosas.   

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report
import numpy as np
import matplotlib.pyplot as plt

y_pred_prob = modelo.predict(X_prueba_CNN)
y_pred_num = np.argmax(y_pred_prob, axis=1)

cm = confusion_matrix(y_prueba_num, y_pred_num)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=codificador.classes_)

fig, ax = plt.subplots(figsize=(7, 6))
disp.plot(cmap='Blues', ax=ax, values_format='d')
plt.title('Matriz de Confusión - Conjunto de Prueba')
plt.show()

print(classification_report(y_prueba_num, y_pred_num, target_names=codificador.classes_))
```

```python
niveles_ordenados = sorted(np.unique(snr_prueba))
precisiones_snr = []

print("Evaluación por nivel de SNR:\n")
for snr in niveles_ordenados:
    indices = np.where(snr_prueba == snr)[0]
    _, acc = modelo.evaluate(X_prueba_CNN[indices], y_prueba_CNN[indices], verbose=0)
    precisiones_snr.append(acc * 100)
    print(f"SNR {snr:>3} dB -> Precisión: {acc*100:.2f}% ({len(indices)} señales)")
```

```python
plt.figure(figsize=(9, 5))
plt.plot(niveles_ordenados, precisiones_snr, marker='o', linewidth=2, color='#1f77b4')
plt.axhline(y=25, color='r', linestyle='--', label='Azar puro (25%)')
plt.title('Precisión del modelo según el nivel de ruido (SNR)')
plt.xlabel('Nivel de SNR (dB)')
plt.ylabel('Precisión (%)')
plt.xticks(niveles_ordenados)
plt.ylim(0, 105)
plt.grid(True)
plt.legend()
plt.show()
```

```python
mascara_ruido_alto = snr_prueba <= 0
cm_ruido_alto = confusion_matrix(y_prueba_num[mascara_ruido_alto], y_pred_num[mascara_ruido_alto])

print("Matriz de confusión (SNR <= 0 dB):\n")
print(cm_ruido_alto)
print("\nPrecisión por modulación con ruido alto (SNR <= 0 dB):")
for i, clase in enumerate(codificador.classes_):
    total_clase = np.sum(cm_ruido_alto[i])
    aciertos = cm_ruido_alto[i, i]
    print(f"{clase}: {aciertos}/{total_clase} ({aciertos/total_clase*100:.2f}%)")
```

Introducción
En los sistemas modernos de telecomunicaciones inalámbricas, redes móviles (4G/5G), Wi-Fi y radio cognitiva, los receptores enfrentan el reto de operar en entornos dinámicos donde el espectro electromagnético es compartido por múltiples servicios y estándares. La Clasificación Automática de Modulaciones (AMC, por sus siglas en inglés) es un proceso intermedio fundamental entre la detección de la señal y su demodulación. Su objetivo consiste en identificar de manera autónoma el esquema de modulación empleado por un transmisor sin contar con información previa sobre sus parámetros de transmisión. Tradicionalmente, la identificación de modulaciones dependía de algoritmos basados en extracción manual de características estadísticas o espectrales y umbrales de decisión complejos, los cuales degradan su desempeño ante canales ruidosos. El uso de técnicas de aprendizaje profundo (Deep Learning), específicamente las Redes Neuronales Convolucionales (CNN), permite procesar directamente las muestras temporales de las señales y aprender automáticamente patrones distintivos de amplitud, frecuencia y fase, logrando sistemas más robustos y adaptables frente a interferencias y ruido del canal.
Metodología
Se implementó una Red Neuronal Convolucional unidimensional (Conv1D) en TensorFlow/Keras, elegida por su capacidad para extraer patrones locales a lo largo de series temporales sin requerir transformaciones previas a espectrogramas. La arquitectura diseñada consta de 38,084 parametros entrenables distribuidos en las siguientes capas:
Capa de entrada: Tensor de dimensión (128, 1) correspondiente a las $128\text{ muestras}$ temporales de cada señal.   
Primera capa convolucional (Conv1D): $32\text{ filtros}$ con tamaño de kernel $3$ y función de activación ReLU para detectar transiciones básicas de amplitud, frecuencia y fase.   
Segunda capa convolucional (Conv1D): $64\text{ filtros}$ con tamaño de kernel $3$ y activación ReLU para extraer características temporales de mayor abstracción.  
Capa de aplanamiento (Flatten): Convierte los mapas de características en un vector unidimensional de 7,936 elementos
Capa de salida (Dense): $4\text{ neuronas}$ con activación Softmax para estimar la distribución de probabilidad sobre las cuatro modulaciones (ASK, FSK, PSK, QAM).   
Resultados
Desempeño Durante el EntrenamientoDurante el entrenamiento, la precisión sobre el conjunto de entrenamiento alcanzó el $100\%$ a partir de la época 7, mientras que la precisión de validación se estabilizó alrededor del $96.1\%$, demostrando una rápida convergencia del modelo.
Respecto a la posible confusión entre ASK y PSK, se observó que con niveles de SNR favorables ambas clases se separan correctamente; sin embargo, ante niveles altos de ruido (SNR <= 0 dB) aparecen confusiones puntuales. Esto ocurre porque ASK codifica la información en cambios de amplitud ($0.3$ vs. $1.0$) y PSK en inversiones de fase ($0$ vs. $\pi$), pero el ruido aditivo AWGN altera la envolvente de la señal y genera cruces irregulares que hacen que un cambio brusco de fase en PSK se asemeje temporalmente a una caída de amplitud en ASK.   
 Análisis de Errores por ModulaciónAl analizar exclusivamente las señales bajo ruido intenso (SNR <= 0 dB), de aciertos. Desde el punto de vista técnico, 16-QAM distribuye 16 estados distintos en el plano $I/Q$, lo que reduce la distancia euclidiana entre símbolos vecinos; al sumarse ruido AWGN intenso, las muestras se desplazan fácilmente hacia regiones de otros símbolos y presentan variaciones de envolvente similares a ASK, dificultando su separación mediante una entrada temporal unidimensional.  
Conclusiones
El desarrollo de “El Detective de Señales” demostró que una arquitectura Conv1D ligera es capaz de clasificar automáticamente modulaciones digitales (ASK, FSK, PSK y 16-QAM) directamente a partir de muestras temporales sin extracción manual de características, alcanzando una precisión superior al $95\%$ en condiciones de SNR positivas (>= 5 dB).
Entre las principales limitaciones observadas se encuentra la degradación de rendimiento en regímenes de SNR negativos (-5 dB), donde el ruido aditivo enmascara las variaciones de amplitud y fase, así como una ligera brecha entre entrenamiento (100) y validación (96) indicativa de sobreajuste. Como trabajo futuro para incrementar la robustez, se propone incorporar capas de regularización (Dropout y MaxPooling1D), ampliar el conjunto de entrenamiento con más muestras en SNR bajos y alimentar la red con dos canales independientes correspondientes a las componentes en fase (I) y cuadratura (Q).
